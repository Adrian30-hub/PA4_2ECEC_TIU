```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_excel("board2.xlsx")
df
```

```python
subs = ['Math', 'Electronics', 'GEAS', 'Communication']
df['Average'] = df[subs].mean(axis=1)

VisComm = pd.DataFrame(df[
                      (df['Hometown'] == 'Visayas') &
        		      (df['Track'] == 'Communication')],
        		      columns = ['Name', 'Gender', 'Math', 'Electronics', 'Average'])
VisComm
```

```python
subs = ['Math', 'Electronics', 'GEAS', 'Communication']
df['Average'] = df[subs].mean(axis=1)

VisFemale = pd.DataFrame(df[
            			(df['Hometown'] == 'Visayas') &
            			(df['Gender'] == 'Female')],
            			columns = ['Name', 'Track', 'GEAS', 'Electronics', 'Average'])
VisFemale
```

```python
rows = VisFemale.loc[(VisFemale['Average'] >= 60), ['Name', 'Track', 'GEAS', 'Electronics', 'Average']]
rows
```

```python
track_avg = df.groupby('Track', as_index=False)['Average'].mean()
gender_avg = df.groupby('Gender', as_index=False)['Average'].mean()
home_avg = df.groupby('Hometown', as_index=False)['Average'].mean()

print("Mean Average by Track")
display(track_avg)

print("Mean Average by Gender")
display(gender_avg)

print("Mean Average by Hometown")
display(home_avg)
```

```python
mean_track = df.groupby('Track')['Average'].mean().reset_index()
print('---Mean Average By Track---')
display(mean_track)

mean_gender = df.groupby('Gender')['Average'].mean().reset_index()
print('---Mean Average By Gender---')
display(mean_gender)

mean_hometown = df.groupby('Hometown')['Average'].mean().reset_index()
print('---Mean Average By Hometown---')
display(mean_hometown)

figure, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

axes[0].bar(mean_track['Track'], mean_track['Average'])
axes[0].set_title('Mean Average By Track')
axes[0].set_xlabel('Track')
axes[0].set_ylabel('Mean Average Scores')

axes[1].bar(mean_gender['Gender'], mean_gender['Average'])
axes[1].set_title('Mean Average By Gender')
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('Mean Average Scores')

axes[2].bar(mean_hometown['Hometown'], mean_hometown['Average'])
axes[2].set_title('Mean Average By Hometown')
axes[2].set_xlabel('Hometown')
axes[2].set_ylabel('Mean Average Scores')

plt.tight_layout()
plt.show()

best_track = mean_track.loc[mean_track['Average'].idxmax()]
best_gender = mean_gender.loc[mean_track['Average'].idxmax()]
best_hometown = mean_hometown.loc[mean_track['Average'].idxmax()]

print(
    f"1. Track: Within the dataset, the track with the highest mean average is {best_track['Track']}, with an average of ({best_track['Average']:.2f})"
) 
print(
    f"2. Gender: Within the dataset, the gender with the highest mean average is {best_gender['Gender']}, with an average of ({best_gender['Average']:.2f})"
) 
print(
    f"3. Hometown: Within the dataset, the hometown with the highest mean average is {best_hometown['Hometown']}, with an average of ({best_hometown['Average']:.2f})"
)
```
